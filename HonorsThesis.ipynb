```python
# Required MRI/ML libraries
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Loads and opens small subset of the selected research dataset with MRI & Segmentation
mri = nib.load(r"C:\Users\annak\Downloads\UCSF-PDGM-0004_DWI.nii.gz").get_fdata()
seg = nib.load(r"C:\Users\annak\Downloads\UCSF-PDGM-0004_brain_segmentation.nii.gz").get_fdata()
```

```python
# Inspect dimensions
print(mri.shape)
print(seg.shape)
```

```python
# Displays middle slice with segmentation overlay
slice_idx = mri.shape[2] // 2
plt.imshow(np.rot90(mri[:,:,slice_idx]), cmap="gray")
plt.imshow(np.rot90(seg[:,:,slice_idx]), cmap="jet", alpha=0.4)
plt.show()
```

```python
# Interactive representation of the MRI image slices
import matplotlib.pyplot as plt
from ipywidgets import interact

def view_slice(i):
    plt.imshow(np.rot90(mri[:,:,i]), cmap="gray")
    plt.imshow(np.rot90(seg[:,:,i]), cmap="jet", alpha=0.4)
    plt.title(f"Slice {i}")
    plt.show()

interact(view_slice, i=(0, mri.shape[2]-1))
```
