```python
# Import required MRI/ML libraries
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Loads and opens a small subset of the selected research dataset
mri = nib.load(r"C:\Users\annak\OneDrive - UNC-Wilmington\HON Thesis\MRIimagesExamples\t1_icbm_normal_1mm_pn3_rf20.mnc").get_fdata()
print("MRI shape:", mri.shape)
```

```python
# Choose slice indexes
sag = mri.shape[0] // 2      # sagittal slice index
cor = mri.shape[1] // 2      # coronal slice index
axi = mri.shape[2] // 2      # axial slice index
```

```python
# Extract slices
sag_slice = mri[sag, :, :]
cor_slice = mri[:, cor, :]
axi_slice = mri[:, :, axi]
```

```python
# Display slices
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(np.rot90(sag_slice), cmap="gray")
ax[0].set_title("Sagittal Slice")
ax[0].axis("off")

ax[1].imshow(np.rot90(cor_slice), cmap="gray")
ax[1].set_title("Coronal Slice")
ax[1].axis("off")

ax[2].imshow(np.rot90(axi_slice), cmap="gray")
ax[2].set_title("Axial Slice")
ax[2].axis("off")

plt.show()
```
